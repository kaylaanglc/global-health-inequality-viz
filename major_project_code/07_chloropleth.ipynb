```python
import pandas as pd
import plotly.express as px

# --------------------------------------------------
# Step 1: Load latest data (after merging with region/income)
# --------------------------------------------------
df = pd.read_csv("cleaned_world_health_data.csv")

# Keep only actual countries
df_country = df[df["entity_type"] == "country"].copy()

# Keep only the most recent year per country
df_latest = df_country.sort_values("year").groupby("country").tail(1)

# Optional: check for missing life expectancy values
df_latest = df_latest.dropna(subset=["life_expect"])

# --------------------------------------------------
# Step 2: Create interactive choropleth with Plotly
# --------------------------------------------------
fig = px.choropleth(
    df_latest,
    locations="country",
    locationmode="country names",
    color="life_expect",
    hover_name="country",
    color_continuous_scale="Viridis",
    title="Geographical Distribution of Life Expectancy (Latest Available Year)",
    labels={"life_expect": "Life Expectancy (Years)"},
    height=600
)

fig.update_geos(showframe=False, showcoastlines=True, projection_type="natural earth")
fig.update_layout(margin={"r":0,"t":50,"l":0,"b":0})

fig.show()
```
