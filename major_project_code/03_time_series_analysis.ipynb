Time-Series Line Plot of Life Expectancy Over Time by Income Group or Region
Since your dataset doesn't include income groups or regions by default, we’ll need to manually map countries to their income groups or regions (using a World Bank reference). For now, we’ll simulate it with a minimal set (you can expand it later).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset
df = pd.read_csv("cleaned_world_health_data.csv")

# -----------------------------
# Define lists of region & income level labels
# -----------------------------
world_regions = [
    "East Asia & Pacific", "Europe & Central Asia", "Latin America & Caribbean",
    "Middle East & North Africa", "North America", "South Asia", "Sub-Saharan Africa"
]

income_levels = [
    "High income", "Upper middle income", "Lower middle income", "Low income",
    "Middle income", "Low & middle income"
]

# -----------------------------
# Plot 1: Life Expectancy by World Region
# -----------------------------
df_region = df[df["country"].isin(world_regions)].copy()
df_region["region"] = df_region["country"]  # create a 'region' column

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_region, x="year", y="life_expect", hue="region", estimator="mean", ci="sd")
plt.title("Life Expectancy Over Time by World Region")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (years)")
plt.legend(title="Region", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()

# -----------------------------
# Plot 2: Life Expectancy by Income Level
# -----------------------------
df_income = df[df["country"].isin(income_levels)].copy()
df_income["income_level"] = df_income["country"]  # create an 'income_level' column

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_income, x="year", y="life_expect", hue="income_level", estimator="mean", ci="sd")
plt.title("Life Expectancy Over Time by Income Level")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (years)")
plt.legend(title="Income Level", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
