```python
### 3D Mesh Plot for Visualizing Health Indicator Interaction
# This Jupyter Notebook cell uses Plotly to create a 3D mesh to visualize how
# health_exp (x-axis), maternal_mortality (y-axis), and life_expect (z-axis)
# interact across countries.

import pandas as pd
import plotly.graph_objects as go
from sklearn.preprocessing import MinMaxScaler

# ---------------------------------------------
# Step 1: Load and prepare the data
# ---------------------------------------------
df = pd.read_csv("cleaned_world_health_data.csv")
df_country = df[df["entity_type"] == "country"].copy()
df_latest = df_country.sort_values("year").groupby("country").tail(1)

# Drop rows with missing values in the selected indicators
plot_df = df_latest.dropna(subset=["health_exp", "maternal_mortality", "life_expect"]).copy()

# Normalize for visual clarity (optional)
scaler = MinMaxScaler()
plot_df[["health_exp", "maternal_mortality", "life_expect"]] = scaler.fit_transform(
    plot_df[["health_exp", "maternal_mortality", "life_expect"]]
)

# Extract values
x = plot_df["health_exp"].values
y = plot_df["maternal_mortality"].values
z = plot_df["life_expect"].values

# ---------------------------------------------
# Step 2: Generate 3D Mesh Plot
# ---------------------------------------------
fig = go.Figure(data=[
    go.Mesh3d(
        x=x,
        y=y,
        z=z,
        color='lightblue',
        opacity=0.5,
        alphahull=6,
        name='Health Surface'
    ),
    go.Scatter3d(
        x=x,
        y=y,
        z=z,
        mode='markers',
        marker=dict(size=4, color='darkblue'),
        text=plot_df["country"],
        name='Countries'
    )
])

fig.update_layout(
    scene=dict(
        xaxis_title='Health Spending (normalized)',
        yaxis_title='Maternal Mortality (normalized)',
        zaxis_title='Life Expectancy (normalized)'
    ),
    title="3D Mesh: Life Expectancy vs Health Spending & Maternal Mortality",
    height=700,
    margin=dict(l=0, r=0, t=40, b=0)
)

fig.show()
```
